# Štiri ozka vprašanja, vnaprej zapisana
### Preverjanje hipotez s statističnimi testi

V prejšnjih zvezkih smo opisali, kaj se je dogajalo z zaposlovanjem in odpuščanji v IT po letu 2020. Tu postavimo **štiri ozka vprašanja**, na katera lahko odgovorimo z »da« ali »ne«, in jih preverimo s statističnimi testi.

**Kako smo poskrbeli, da je odgovor pošten:**
- **Vse smo zapisali vnaprej.** Preden smo pognali kateri koli test, smo v datoteko `docs/hypotheses/PREREGISTRATION.md` zapisali hipoteze, podatke, en glavni test za vsako, mejo za odločitev in kaj smo že vedeli. Datoteka je bila shranjena v git 14. 11. 2026 ob 9.05, pred prvim izračunom. Rezultata tako ne moremo »prilagoditi« podatkom.
- **Vsak test smo najprej preizkusili na izmišljenih podatkih**: kjer smo učinek podtaknili, ga je moral najti, kjer ga ni bilo, ga ni smel.
- **Vsak test ima »placebo«**: isti postopek tam, kjer učinka ne bi smelo biti. Pokaže, kako pogosto test zmotno najde učinek (pričakujemo okoli 5 %).
- **Popravek za več testov.** Če naredimo štiri teste, je verjetnost, da bo vsaj eden po naključju »uspel«, večja kot pri enem. Zato p-vrednosti popravimo po metodi Holm.

**Kako brati:** p-vrednost pove, kako verjeten bi bil tako velik učinek, če v resnici učinka ne bi bilo. Hipoteza je **potrjena** le, če je p po popravku manjši od 0,05 **in** je učinek v napovedani smeri. »Ni potrjena« ne pomeni, da učinka zagotovo ni, ampak da ga naši podatki ne morejo dokazati; interval pove, kako velik bi lahko bil. Pod vsakim grafom so štiri razlage (**Kaj vidimo?**, **Zakaj smo to naredili?**, **Kaj to pomeni?**, **Česa ne vemo?**) in vrstica **Viri podatkov**.

Eno vprašanje smo opustili že pred testom: ali padec prihodkov napove odpuščanja v podjetjih s polprevodniki bolj kot v programski opremi. Takih podjetij je bilo premalo (le 10), da bi test lahko dal zanesljiv odgovor.

```python
from pathlib import Path
import os, sys
from IPython.display import Markdown, display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "build_master.py").exists())
sys.path.insert(0, str(ROOT)); os.chdir(ROOT)          # poti do podatkov so relativne glede na mapo projekta
import plotly.io as pio
from analysis import hypotheses_viz as hv
from analysis.viz_theme import explanation_markdown
from analysis.data_sources import sources_markdown

# Grafi se vidijo v JupyterLab in v izvoženem HTML (ta potrebuje internet, da naloži plotly.js).
pio.renderers.default = "plotly_mimetype+notebook_connected"

results = hv.load_results()                            # številke iz enkratnega, vnaprej zapisanega izračuna
data = hv.prepare_figure_data()                        # podatki za grafe (isti vzorci kot v testih)
items = hv.build_hypotheses(results, data)


def show(name, missing_text):
    """Graf, pod njim štiri razlage v preprostem jeziku in vrstica z viri podatkov."""
    if name not in items:
        display(Markdown(missing_text))
        return
    fig, parts = items[name]
    fig.show()
    display(Markdown(explanation_markdown(parts) + "\n\n" + sources_markdown(name)))
```

## Česa ti testi ne povedo

- **Povezava ni vzrok.** Tudi potrjena hipoteza pove le, da se nekaj zgodi skupaj, ne zakaj.
- **»Ni potrjena« ni »dokazano ne«.** Pri majhnih vzorcih lahko test spregleda resničen, a majhen učinek; zato vedno navajamo interval.
- **Teste smo izvedli enkrat.** Če bi po rezultatu spreminjali vzorec ali test, bi p-vrednosti izgubile pomen; vse spremembe bi morali zapisati kot dopolnitev registracije.
